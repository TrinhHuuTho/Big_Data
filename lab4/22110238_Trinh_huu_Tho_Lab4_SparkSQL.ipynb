## 5. Câu hỏi

Viết truy vấn SQL để trả lời các câu hỏi sau đồng thời thực thi các truy vấn đó để hiển thị kết quả.

### Q1. Liệt kê `name`, `address`, `city`, và `region` của các `employee` sống ở `'USA'`.

```python
query = """
    select firstName, lastName, address, city, region
    from employee
    where Country = 'USA'
"""
spark.sql(query).show(n=10, truncate=50)
```

### Q2. Liệt kê `title` và `name` của các `employee` đã bán ít nhất một `product` có tên `'Gravad Lax'` hoặc `'Mishi Kobe Niku'`.

```python
query = """
    select distinct
        e.Title, 
        concat(e.FirstName ,' ', e.LastName) AS EmployeeName
    from employee as e
    join order as o
        on e.Id = o.EmployeeId
    join orderdetail as od
        on o.Id = od.OrderId
    join product as p
        on od.ProductId = p.Id
    where p.ProductName in ('Gravad Lax', 'Mishi Kobe Niku')
"""
spark.sql(query).show(n=10, truncate=50)
```

### Q3. Liệt kê `name` và `title` của các `employee` và `name` và `title` của các `employee` khác mà họ phải báo cáo (`ReportsTo`) (hoặc `null` nếu họ không phải báo cáo với ai). 

```python
query = """
    select e.Title as EmployeeTitle, e.FirstName as EmployeeFirstName,
            m.Title as ManagerTitle, m.FirstName as ManagerFirstName
    from employee as e  
    left join employee as m
        on e.ReportsTo = m.Id          
"""
spark.sql(query).show(n=10, truncate=100)
```

### Q4. Liệt kê `customer name`, `product name` và `supplier name` cho các `customer` sống ở `'London'` và các `supplier` có tên `'Pavlova, Ltd.'` hoặc `'Karkki Oy'`.

```python
query = """
    select 
        c.CompanyName AS CustomerName,
        p.ProductName,
        s.CompanyName
        
    from customer as c
    join order as o
        on c.Id = o.CustomerId
    join orderdetail as od
        on o.Id = od.OrderId
    join product as p
        on od.ProductId = p.Id
    join supplier as s
        on p.SupplierId = s.Id
    where c.City = 'London' and s.CompanyName in ('Pavlova, Ltd.', 'Karkki Oy')
"""
spark.sql(query).show(n=20, truncate=60)
```

### Q5. Liệt kê `name` của các `employee` có tuổi lớn hơn **một `employee` nào đó** sống ở `'London'`.

```python
query = """
    select e.FirstName, e.LastName
    from employee e
    where exists(
        select 1
        from employee l
        where l.City = 'London' and l.BirthDate > e.BirthDate
    )
"""
spark.sql(query).show(n=10)
```

### Q6. Liệt kê `name` của các `employee` có tuổi lớn hơn **tất cả** `employee` sống ở `'London'`.

```python
query = """
    select e.FirstName, e.LastName
    from employee e
    where e.BirthDate < (
        select min(l.BirthDate)
        from employee l
        where l.City = 'London'
    )
"""
spark.sql(query).show(n=10)
```

### Q7. Liệt kê `name` của các `employee` làm việc lâu năm hơn **tất cả** `employee` ở `'London'`.

### Q8. Liệt kê name của các `employee` và `city` nơi họ sống đối với các `employee` đã bán hàng cho các `customer` ở cùng thành phố với họ.

```python
query = """
    select distinct e.FirstName, e.LastName, e.City // dissinct de loai bo cac dong trung lap
    from employee e
    join order o
        on e.Id = o.EmployeeId
    join customer c
        on o.CustomerId = c.Id
    where e.City = c.City
"""
spark.sql(query).show(n=10, truncate=50)
```

### Q10. Liệt kê `name` của các `category` và giá trung bình (`average price`) của các `product` trong mỗi `category`.

```python
query = """
    select c.CategoryName, avg(p.UnitPrice) as AveragePrice
    from category c
    join product p
        on c.Id = p.CategoryId
    group by c.CategoryName
    order by c.CategoryName 
"""
spark.sql(query).show(n=10, truncate=50)
```

### Q11. Liệt kê `id` và `name` của các `supplier` cung cấp hơn 3 `product`.

```python
query = """
    select
        s.Id,
        s.CompanyName
    from supplier s
    join product p
        on s.Id = p.SupplierId
    group by s.Id, s.CompanyName
    having count(p.Id) > 3
"""
spark.sql(query).show(n=20, truncate=False)
```

### Q12. Liệt kê `id`, `name`, và `số lượng order` của các `employee` sắp xếp theo `employee id`.

### Q13. Với mỗi `employee`, liệt kê `id`, `name`, và `số loại product` khác nhau họ đã bán, sắp xếp `employee id`.

```python
query = """
    select
        e.Id,
        e.FirstName,
        e.LastName,
        count(distinct p.Id) AS ProductNum
    from employee e
    join order o 
        on e.Id = o.EmployeeId
    join orderdetail od 
        on o.Id = od.OrderId
    join product p 
        on od.ProductId = p.Id
    group by e.Id, e.FirstName, e.LastName
    order by e.Id
"""
spark.sql(query).show(n=20, truncate=False)
```

### Q14. Liệt kê `id`, `name`, và `tổng doanh thu` (sau khi trừ đi chiết khấu - `Discount`) của mỗi `employee`, sắp xếp theo `employee id`.

```python
query = """
    select
        e.Id,
        e.FirstName,
        e.LastName,
        sum(od.UnitPrice * od.Quantity * (1 - od.Discount)) as TotalRevenue
    from employee e
    join order o
        on e.Id = o.EmployeeId
    join orderdetail od
        on o.Id = od.OrderId
    group by
        e.Id,
        e.FirstName,
        e.LastName
    order by
        e.Id
"""
spark.sql(query).show(n=20, truncate=False)
```

### Q15. Liệt kê `id`, `name`, và `tổng doanh thu` (có trừ đi chiết khấu - `Discount`) của các `employee` đã bán được hơn `70` `product` khác nhau, sắp xếp theo `employee id`.

```python
query = """
    select
        e.Id as Id,
        e.FirstName,
        e.LastName,
        -- Tính tổng doanh thu sau chiết khấu
        sum(od.UnitPrice * od.Quantity * (1 - od.Discount)) as TotalRevenue
    from employee e
    join order o
        on e.Id = o.EmployeeId
    join orderdetail od
        on o.Id = od.OrderId
    group by
        e.Id,
        e.FirstName,
        e.LastName
    having
        -- Điều kiện: đếm số lượng ProductId khác nhau > 70
        count(distinct od.ProductId) > 70
    order by
        e.Id
"""
spark.sql(query).show(n=20, truncate=False)
```

### Q16. Liệt kê `name` của các `employee` đã bán các `product` của hơn `28` `supplier`.

```python
query = """
    select
        e.Id,
        e.FirstName,
        e.LastName
    from employee e
    join order o
        on e.Id = o.EmployeeId
    join orderdetail od
        on o.Id = od.OrderId
    join product p
        on od.ProductId = p.Id
    join supplier s
        on p.SupplierId = s.Id -- Nối sản phẩm với nhà cung cấp
    group by
        e.Id,
        e.FirstName,
        e.LastName
    having
        -- Đếm số lượng SupplierId khác nhau mà employee đã bán sản phẩm
        count(distinct s.Id) > 28
"""
spark.sql(query).show(n=20, truncate=False)
```

### Q17. Liệt kê `customer name` và `product name` sao cho `số lượng product` đó được mua bởi `customer` đó trong một `order` lớn hơn `5` lần `số lượng trung bình` `product` đó được mua trong một `order` bởi **tất cả** các `customer`.

```python
query = """
    with avgqty as (
        -- 1. Tính số lượng trung bình (AVG(Quantity)) cho MỖI sản phẩm (ProductId)
        select
            od.ProductId,
            avg(od.Quantity) as GlobalAvg
        from orderdetail od
        group by od.ProductId
    )

    select distinct
        c.ContactName,
        p.ProductName
    from customer c
    join order o
        on c.Id = o.CustomerId
    join orderdetail od
        on o.Id = od.OrderId
    join product p
        on od.ProductId = p.Id
    join avgqty a
        on p.Id = a.ProductId -- Nối với bảng trung bình toàn cục
    where
        -- 2. Lọc các chi tiết đơn hàng thỏa mãn điều kiện
        od.Quantity > (5 * a.GlobalAvg)
"""
spark.sql(query).show(n=20, truncate=False)
```

### Q18. Liệt kê `name` của các `customer` đã mua **tất cả** `product` với giá `< 5`.

```python
query = """
    select
        c.ContactName
    from customer c
    join order o
        on c.Id = o.CustomerId
    join orderdetail od
        on o.Id = od.OrderId
    join product p
        on od.ProductId = p.Id
    where
        -- Lọc để chỉ xử lý các sản phẩm có giá < 5
        p.UnitPrice < 5
    group by
        c.Id,
        c.ContactName
    having
        -- Điều kiện Phép chia:
        -- Số lượng ProductId khác nhau mà customer này đã mua (với giá < 5)
        count(distinct p.Id) = (
            -- Bằng TỔNG số ProductId khác nhau trong toàn bộ CSDL có giá < 5
            select count(Id) from product where UnitPrice < 5
        )
"""
spark.sql(query).show(n=20, truncate=False)
```

### Q19. Liệt kê `name` của các `product` được bán bởi **mọi** `employee`.

```python
query = """
    select
        p.ProductName
    from product p
    join orderdetail od
        on p.Id = od.ProductId
    join order o
        on od.OrderId = o.Id
    group by
        p.Id,
        p.ProductName
    having
        -- Điều kiện Phép chia:
        -- Đếm số lượng EmployeeId khác nhau đã bán sản phẩm này
        count(distinct o.EmployeeId) = (
            -- Bằng tổng số lượng Employee trong bảng employee
            select count(Id) from employee
        )
"""
spark.sql(query).show(n=20, truncate=False)
```

### Q20. Liệt kê `name` của các `customer` đã mua **tất cả** các `product` được đặt hàng bởi `customer` có `id` là `'LAZYK'`.

```python
query = """
    select
        c.ContactName
    from customer c
    join order o
        on c.Id = o.CustomerId
    join orderdetail od
        on o.Id = od.OrderId
    where
        -- Lọc để chỉ xem xét các ProductId đã từng được mua bởi customer 'LAZYK'
        od.ProductId in (
            select od_lazyk.ProductId
            from orderdetail od_lazyk
            join order o_lazyk on od_lazyk.OrderId = o_lazyk.Id
            where o_lazyk.CustomerId = 'LAZYK'
        )
    group by
        c.Id,
        c.ContactName
    having
        -- Điều kiện Phép chia:
        -- Số lượng ProductId khác nhau mà customer này đã mua (thuộc tập sản phẩm của LAZYK)
        count(distinct od.ProductId) = (
            -- Bằng tổng số lượng ProductId khác nhau mà LAZYK đã mua
            select count(distinct T1.ProductId)
            from orderdetail T1
            join order T2 on T1.OrderId = T2.Id
            where T2.CustomerId = 'LAZYK'
        )
"""
spark.sql(query).show(n=20, truncate=False)
```
